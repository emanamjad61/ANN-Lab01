# Task 4 — Indexing and Slicing Practice

*Topics Covered: Indexing and Slicing*

**Part (a)** — Create a list of the numbers 1 to 15 using `range()` and `list()`.

```python
# Task 4(a)  |  Eman Amjad  |  FA23-BAI-013
numbers = list(range(1, 16))

print("The list :", numbers)
print("Length   :", len(numbers))
print("Starts at", numbers[0], "and ends at", numbers[-1])

# range(1, 16) is exclusive of its stop value, which is why we write 16 to reach 15
```

**Part (b)** — Extract and print the first 4 elements and the last 4 elements using slicing.

```python
# Task 4(b)  |  Eman Amjad  |  FA23-BAI-013
first_four = numbers[0:4]     # from index 0 up to (not including) index 4
last_four = numbers[11:15]    # the final four positions

print("Whole list        :", numbers)
print("First four [0:4]  :", first_four)
print("Last four [11:15] :", last_four)

# Negative indexing is the shorter way to reach the end of a list
print("Last four [-4:]   :", numbers[-4:], "<- same result, cleaner")
```

**Part (c)** — Extract and print every 3rd element of the list starting from index 0.

```python
# Task 4(c)  |  Eman Amjad  |  FA23-BAI-013
step_three = numbers[0::3]     # start : stop : step

print("Whole list          :", numbers)
print("Every 3rd element   :", step_three)
print("How many were taken :", len(step_three))

# The third value in a slice is the STEP, so it jumps forward 3 places each time
```

**Part (d)** — Print the list in reverse order using slicing (without using the `reverse()` function).

```python
# Task 4(d)  |  Eman Amjad  |  FA23-BAI-013
backwards = numbers[::-1]      # a step of -1 walks from the end to the start

print("Forward  :", numbers)
print("Backward :", backwards)

# Slicing never edits the list it came from, it builds a brand new one
print("\nIs the original still in order?", numbers == list(range(1, 16)))
```

**Part (c)** — Write nested `if` statements to check whether a given year is a leap year.

```python
# Task 6(c)  |  Eman Amjad  |  FA23-BAI-013
year = 1900

# Rule: divisible by 4 -> leap, BUT century years must also be divisible by 400
if year % 4 == 0:
    if year % 100 == 0:              # it is a century year
        if year % 400 == 0:
            print(f"{year} IS a leap year (century year divisible by 400)")
        else:
            print(f"{year} is NOT a leap year (century year not divisible by 400)")
    else:
        print(f"{year} IS a leap year")
else:
    print(f"{year} is NOT a leap year (not divisible by 4)")
```

# Task 7 — Loops with Break and Continue

*Topics Covered: Flow Control (`for` / `while` / `break` / `continue`)*

**Part (a)** — Use a `for` loop to print all numbers from 1 to 30, but skip every multiple of 4 using `continue`.

```python
# Task 7(a)  |  Eman Amjad  |  FA23-BAI-013
printed = []
skipped = []

for value in range(1, 31):
    if value % 4 == 0:
        skipped.append(value)
        continue        # jumps straight to the next value, skipping the code below
    printed.append(value)

print("Numbers printed:")
print(printed)

print("\nNumbers skipped (multiples of 4):")
print(skipped)

print(f"\n{len(printed)} printed, {len(skipped)} skipped, {len(printed) + len(skipped)} total")
```

**Part (b)** — Use a `for` loop to print numbers from 1 to 50, and stop the loop as soon as a number greater than 20 is found, using `break`.

```python
# Task 7(b)  |  Eman Amjad  |  FA23-BAI-013
for value in range(1, 51):
    if value > 20:
        print(f"\n\n{value} is greater than 20 - stopping the loop here with break")
        break           # ends the loop completely, the remaining values are never reached
    print(value, end="  ")

print(f"The loop was set up to go to 50 but only reached {value}.")
```

**Part (c)** — Use a `while` loop to print the multiplication table (1 to 10) of a number provided in a variable.

```python
# Task 7(c)  |  Eman Amjad  |  FA23-BAI-013
table_of = 9
counter = 1

print(f"--- Multiplication table of {table_of} ---")

while counter <= 10:
    product = table_of * counter
    print(f"{table_of} x {counter:2d} = {product:3d}")
    counter += 1        # the loop would run forever without this line

print("--- end of table ---")
```

**Part (d)** — Write a loop that calculates and prints the sum of all even numbers between 1 and 100.

```python
# Task 7(d)  |  Eman Amjad  |  FA23-BAI-013
running_total = 0
current = 1

# A while loop this time, to show the same job done a different way
while current <= 100:
    if current % 2 == 0:
        running_total += current
    current += 1

print("Sum of all even numbers from 1 to 100 =", running_total)

# Quick sanity check using a different method entirely
check = sum(range(2, 101, 2))
print("Checked against sum(range(2, 101, 2)) =", check)
print("Do both methods agree?", running_total == check)
```

# Task 9 — Classes and Inheritance

*Topics Covered: Classes (OOP Basics)*

**Part (a)** — Create a class `Course` with an `__init__` method that stores `course_name` and `credit_hours`, and a method `display_info()` that prints these details.

```python
# Task 9(a)  |  Eman Amjad  |  FA23-BAI-013
class Course:
    """A single university course."""

    def __init__(self, course_name, credit_hours):
        # __init__ runs automatically whenever a new object is created
        self.course_name = course_name
        self.credit_hours = credit_hours

    def display_info(self):
        print(f"Course       : {self.course_name}")
        print(f"Credit Hours : {self.credit_hours}")


print("The Course class is ready to use.")
```

**Part (b)** — Create two objects of the `Course` class with different data and call `display_info()` on both.

```python
# Task 9(b)  |  Eman Amjad  |  FA23-BAI-013
database_course = Course("Database Systems", 4)
stats_course = Course("Probability and Statistics", 3)

print("Object 1")
database_course.display_info()

print("\nObject 2")
stats_course.display_info()

# Both objects come from one class but keep their own separate data
print("\nAre they the same object?", database_course is stats_course)
```

**Part (c)** — Create a class `LabCourse` that inherits from `Course` and adds an extra attribute `lab_hours`, using `super().__init__()` to reuse the parent constructor.

```python
# Task 9(c)  |  Eman Amjad  |  FA23-BAI-013
class LabCourse(Course):
    """A course that also has a weekly lab session."""

    def __init__(self, course_name, credit_hours, lab_hours):
        # super() hands the shared work back to Course.__init__
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours


networks_lab = LabCourse("Computer Networks Lab", 1, 3)

print("Inherited from Course -> ", networks_lab.course_name, "|", networks_lab.credit_hours)
print("Added by LabCourse    -> ", networks_lab.lab_hours, "lab hours")

print("\nLabCourse inherits from Course?", issubclass(LabCourse, Course))
print("networks_lab counts as a Course?", isinstance(networks_lab, Course))
```

**Part (d)** — Override the `display_info()` method in `LabCourse` so that it prints all the parent's details plus the `lab_hours`, and demonstrate it by creating one `LabCourse` object.

```python
# Task 9(d)  |  Eman Amjad  |  FA23-BAI-013
class LabCourse(Course):
    """LabCourse with its own version of display_info()."""

    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours

    def display_info(self):
        # Reuse the parent's printing, then add the extra line
        super().display_info()
        print(f"Lab Hours    : {self.lab_hours}")


ann_lab = LabCourse("Artificial Neural Networks Lab", 1, 3)

print("LabCourse object (overridden method):")
ann_lab.display_info()

print("\nCourse object (original method):")
stats_course.display_info()

# Same method name, different behaviour depending on the class - that is overriding
```
