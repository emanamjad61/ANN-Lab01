```python
import numpy as np
```

```python
ages = np.array([25, 32, 41, 19, 55, 47])
print(ages.shape, ages.ndim, ages.dtype)

years_until_retirement = 65 - ages
print(years_until_retirement)

print(np.mean(ages))
```

```python
sales = np.array([
    [120, 150, 130],
    [200, 180, 210],
    [90, 95, 100],
    [300, 280, 320]
])
print(sales.shape, sales.size)

zeros_matrix = np.zeros((4, 3))
zeros_matrix[1] = [10, 20, 30]
print(zeros_matrix)
```

```python
term1 = np.array([
    [70, 65, 80],
    [55, 60, 75],
    [88, 90, 85]
])
term2 = np.array([
    [75, 70, 78],
    [65, 68, 80],
    [90, 92, 89]
])

total_marks = term1 + term2
print(total_marks)

improvement = term2 - term1
print(improvement)

student_totals = improvement.sum(axis=1)
best_student = np.argmax(student_totals)
print(best_student, student_totals)
```

```python
prices = np.array([50, 20, 15, 100])
quantities = np.array([2, 5, 3, 1])

total_bill = np.dot(prices, quantities)
print(total_bill)

customers = np.array([
    [2, 5, 3, 1],
    [1, 0, 4, 2]
])
customer_bills = customers.dot(prices)
print(customer_bills)

# The number of columns in customers (4) must match the number of elements in prices (4)
# because the dot product sums element-wise products along that shared dimension.
```

```python
hours_studied = np.array([2, 3, 1, 4, 5, 2, 3, 4, 1, 2, 3, 5])

reshaped_4x3 = hours_studied.reshape(4, 3)
reshaped_other = hours_studied.reshape(2, 6)
print(reshaped_4x3)
print(reshaped_other)

try:
    hours_studied.reshape(5, 3)
except ValueError as e:
    print(e)
```

```python
matrix = np.random.randint(1, 51, size=(5, 5))
print(matrix)

print(matrix[2, 3])

print(matrix[-1, :])
print(matrix[:, -1])

matrix[0, 0] = 999
print(matrix)
```

```python
full_matrix = np.arange(1, 37).reshape(6, 6)
print(full_matrix)

top_left = full_matrix[:3, :3]
print(top_left)

alternate = full_matrix[::2, ::2]
print(alternate)

last_2x2 = full_matrix[-2:, -2:]
print(last_2x2)
```

```python
np.random.seed(7)
decimal_matrix = np.random.rand(3, 3)
print(decimal_matrix)

int_matrix = np.random.randint(10, 101, size=(3, 3))
print(int_matrix)

normal_matrix = np.random.randn(4, 4)
print(normal_matrix)

np.random.seed(7)
run1 = np.random.rand(3, 3)
np.random.seed(7)
run2 = np.random.rand(3, 3)
print(np.array_equal(run1, run2))

# Both outputs are identical because resetting the seed to the same value
# restarts the pseudo-random generator from the same internal state.
```

```python
performance = np.array([
    [80, 85, 90, 75, 88],
    [70, 72, 68, 74, 71],
    [95, 92, 97, 90, 93]
])

performance_t = performance.T
print(performance_t.shape)

print(np.array_equal(performance.T.T, performance))

a = np.arange(8).reshape(4, 2)
b = np.arange(6).reshape(3, 2)
result = a.dot(b.T)
print(result)
```
